```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.kdeplot(data=df,x="Deaths", fill=True)
plt.title("Death Distribution")
plt.xlabel("Deaths")
plt.ylabel("Density")
plt.show()
```

```python
sns.kdeplot(data=df,x="New recovered", fill=True)
plt.title("New Recovered Distribution")
plt.xlabel("New recovered")
plt.ylabel("Density")
plt.show()
```

```python
sns.kdeplot(data=df,x="Recovered", fill=True)
plt.title("Recovered Distribution")
plt.xlabel("Recovered")
plt.ylabel("Density")
plt.show()
```

```python
sns.violinplot(data=df, x="Country/Region",y="Recovered")
plt.title("Recovered By Country and Region")
plt.shpo
```
