# Matplotlib:For making for visualization with data

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[100,300,250,450,500]
```

# Line chart

```python
plt.plot(x,y, color="red")
plt.title("Sales Report")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

# Bar Graph

```python
x=[1,2,3,4,5,6]
y=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
```

```python
plt.bar(x,y)
plt.title("Day Wise Sales")
plt.show()
```

# horizontal Bar Graph

```python
plt.barh(x,y)
plt.title("Day Wise Sales")
plt.show()
```

# Pie chart

```python
x=[1,2,3,4,5,6]
y=["A","B","C","D","E","F"]
```

```python
plt.pie(x, labels=y,autopct="%1.1f%%")
plt.show()
```

# Histogram

```python
x=[1,2,4,2,4,2,8]
plt.hist(x, bins = 10)
plt.plot()
```

# Scatter Plot

```python
x=[1,2,4,1,10,5,7]
y=[100,200,400,250,188,225,223]
plt.scatter(x,y)
plt.plot()
```

# Box Plot

```python
# it is used for detecting outliers in the data.
```

```python
x=[1,2,1,4,12,32,16,2,13,9,8,280,220,300,340]  #mean= 82.67
plt.boxplot(x)    #median= 12
plt.show()
```

```python
#Methods to remove outliers
# IQR Method   #Z Square method
# First Quartile=2
# Second Quartile= 12
#Third quartile= 220
q1,q2,q3 =2,12,220
```

```python
q1=2
q2=12
q3=220
iqr=q3-q1
#Lower Bound
#Upper Bound
LB=2−1.5 × iqr

UB=230+ 1.5 × iqr
```

# Area Chart

```python
x=[1,2,4,1,10,5,7]
y=[100,200,400,250,188,225,223]
plt.fill_between(x,y)
plt.show()
```

# Stack Chart

```python
x=[1,2,3,4,5,6]
y1=[1,2,4,6,7,8]
y2=[2,3,4,5,6,10]
plt.stackplot(x,y1,y2)
plt.show()
```

```python
x=[0,2,3,4,5,6]
y1=[1,2,4,6,7,8]
y2=[2,3,4,5,6,10]
y3=[0,5,7,9,12,14,]
plt.stackplot(x,y1,y2,y3)
plt.show()
```
